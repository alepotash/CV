```python
from docx import Document
from docx.shared import Inches, Pt, RGBColor
from docx.oxml.ns import nsdecls
from docx.oxml import parse_xml
from docx.enum.text import WD_ALIGN_PARAGRAPH

# --- SETUP ---
document = Document()

# Set Narrow Margins (0.5 inch)
sections = document.sections
for section in sections:
    section.top_margin = Inches(0.5)
    section.bottom_margin = Inches(0.5)
    section.left_margin = Inches(0.5)
    section.right_margin = Inches(0.5)

# Create Layout Table (1 Row, 2 Columns)
table = document.add_table(rows=1, cols=2)
table.autofit = False 
table.columns[0].width = Inches(2.2) # Left Column (Sidebar)
table.columns[1].width = Inches(5.3) # Right Column (Main)

# --- COLOR PALETTE (The "Soft" Adjustments) ---
# Soft Eucalyptus / Muted Sage Green
# This color is desaturated (greyish-green) to remove "aggression"
SIDEBAR_BG_COLOR = "6B8E85"  

# Matching text color for headings (same as sidebar but in RGB)
MAIN_HEADING_COLOR = RGBColor(107, 142, 133) 

# Soft Black for body text (Pure black #000000 can be harsh on screens)
TEXT_COLOR = RGBColor(70, 70, 70) 

# --- HELPER: BACKGROUND COLOR ---
def set_cell_background(cell, fill_color):
    """Sets hex color background for a cell."""
    shading_elm = parse_xml(r'<w:shd {} w:fill="{}"/>'.format(nsdecls('w'), fill_color))
    cell._tc.get_or_add_tcPr().append(shading_elm)

# --- HELPER: SIDEBAR TEXT ---
def add_sidebar_text(cell, text, bold=False, size=10, is_header=False):
    p = cell.add_paragraph()
    run = p.add_run(text)
    run.font.color.rgb = RGBColor(255, 255, 255) # White text
    run.font.name = 'Segoe UI' # Clean, humanistic font
    
    if is_header:
        run.bold = True
        run.font.size = Pt(11) # Slightly smaller header for less "boldness"
        p.space_before = Pt(14) 
        p.space_after = Pt(4)
    else:
        run.bold = bold
        run.font.size = Pt(size)
        p.space_after = Pt(2)

# --- HELPER: MAIN BODY TEXT ---
def add_main_text(cell, text, bold=False, size=10, is_heading=False, color=None):
    p = cell.add_paragraph()
    run = p.add_run(text)
    run.font.name = 'Segoe UI'
    
    if is_heading:
        run.font.size = Pt(14)
        run.bold = True
        if color:
            run.font.color.rgb = color
        p.space_before = Pt(14)
        p.space_after = Pt(6)
    else:
        run.font.size = Pt(size)
        run.bold = bold
        run.font.color.rgb = TEXT_COLOR # Soft Dark Grey

# ==========================================
# LEFT COLUMN (SIDEBAR) - Soft Eucalyptus
# ==========================================
left_cell = table.cell(0, 0)
set_cell_background(left_cell, SIDEBAR_BG_COLOR)

# Spacer
left_cell.add_paragraph() 

# 1. Contact Info
add_sidebar_text(left_cell, "054-3482954", size=10)
add_sidebar_text(left_cell, "alepotash@gmail.com", size=10)
add_sidebar_text(left_cell, "Tel Aviv-Yafo", size=10)
add_sidebar_text(left_cell, "github.com/alepotash", size=10)

# 2. Education
add_sidebar_text(left_cell, "EDUCATION", is_header=True)
add_sidebar_text(left_cell, "Data Analyst Certification", bold=True)
add_sidebar_text(left_cell, "Hebrew University (2025)")
add_sidebar_text(left_cell, "Hands-on training in data analysis & visualization.", size=9)

add_sidebar_text(left_cell, "B.A. Phil & Soc", bold=True)
add_sidebar_text(left_cell, "Hebrew University (2018)")

# 3. Technical Skills
add_sidebar_text(left_cell, "TECHNICAL SKILLS", is_header=True)
add_sidebar_text(left_cell, "Analysis & Viz:", bold=True)
add_sidebar_text(left_cell, "Python, SQL, Qlik, Tableau, Excel (Advanced)")

add_sidebar_text(left_cell, "Operational Systems:", bold=True) 
add_sidebar_text(left_cell, "Priority, ZV")

add_sidebar_text(left_cell, "General:", bold=True)
add_sidebar_text(left_cell, "MS Office, GitHub")

add_sidebar_text(left_cell, "Soft Skills:", bold=True)
add_sidebar_text(left_cell, "Analytical thinking, Organization, Process leadership")

# 4. Languages
add_sidebar_text(left_cell, "LANGUAGES", is_header=True)
add_sidebar_text(left_cell, "Hebrew: Native")
add_sidebar_text(left_cell, "English: Excellent")
add_sidebar_text(left_cell, "Russian: Good")


# ==========================================
# RIGHT COLUMN (MAIN CONTENT)
# ==========================================
right_cell = table.cell(0, 1)

# 1. Name and Title
name_p = right_cell.add_paragraph()
name_run = name_p.add_run("ALEXANDRA POTASHNIKOV")
name_run.bold = True
name_run.font.size = Pt(24) 
name_run.font.name = 'Segoe UI'
name_run.font.color.rgb = MAIN_HEADING_COLOR 

title_p = right_cell.add_paragraph()
title_run = title_p.add_run("Data Analyst")
title_run.font.size = Pt(14)
title_run.font.color.rgb = RGBColor(120, 120, 120) # Soft Grey
title_p.space_after = Pt(12)

# 2. Summary
add_main_text(right_cell, "PROFESSIONAL SUMMARY", is_heading=True, color=MAIN_HEADING_COLOR)
add_main_text(right_cell, "Junior Data Analyst with extensive experience in operational management and budget control. Proven ability to transform business data into operational insights and actionable recommendations. Skilled in building and managing reports for diverse sectors to improve performance, service quality, and profitability.")

# 3. Experience
add_main_text(right_cell, "PROFESSIONAL EXPERIENCE", is_heading=True, color=MAIN_HEADING_COLOR)

# -- Fattal --
add_main_text(right_cell, "Fattal Hotels | Regional Cost Controller", bold=True, size=11)
add_main_text(right_cell, "2024 – Present", size=9)
add_main_text(right_cell, "• Managed budget control and profitability analysis for 8 regional hotels using advanced Excel models.")
add_main_text(right_cell, "• Identified budget anomalies in real-time and presented data-driven insights to senior management.")
add_main_text(right_cell, "• Created specialized reports for ad-hoc projects aimed at performance optimization, utilizing Cursor to accelerate data processing.")

# -- Brown --
add_main_text(right_cell, "\nBrown Hotels | Various Management Roles", bold=True, size=11)
add_main_text(right_cell, "2017 – 2024", size=9)

# Role 1
add_main_text(right_cell, "Network Controller:", bold=True)
add_main_text(right_cell, "• Managed operational control for the entire hotel chain, leading process improvement initiatives.")

# Role 2
add_main_text(right_cell, "Network Service Manager:", bold=True)
add_main_text(right_cell, "• Developed service quality analysis methods; built reports and formulated actionable recommendations regarding customer satisfaction.")
add_main_text(right_cell, "• Led knowledge transfer to operational teams based on data insights.")

# Role 3
add_main_text(right_cell, "Hotel Operations Manager:", bold=True)
add_main_text(right_cell, "• Managed daily hotel operations and conducted Quality Control (QC) during pre-opening stages, working directly with contractors.")

# Previous Roles
p_prev = right_cell.add_paragraph()
p_prev.space_before = Pt(8)
run_prev = p_prev.add_run("Previous Roles: Front Office Manager, Sales & Service Representative, Regional Reservations Manager.")
run_prev.font.size = Pt(9)
run_prev.font.italic = True
run_prev.font.name = 'Segoe UI'
run_prev.font.color.rgb = RGBColor(100, 100, 100) # Lighter grey

# --- SAVE ---
document.save('Alexandra_Potashnikov_CV.docx')
```
